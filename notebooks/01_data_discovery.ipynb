```python
# ============================================================
# CELL 4: INSPECT OFFICIAL SEN1FLOODS11 PUBLIC BUCKET
# ============================================================

import requests
import pandas as pd
from collections import Counter
from pathlib import PurePosixPath

print("=" * 70)
print("SEN1FLOODS11 PUBLIC BUCKET INSPECTION")
print("=" * 70)

# ------------------------------------------------------------
# 1. Public Google Cloud Storage bucket configuration
# ------------------------------------------------------------

BUCKET_NAME = "sen1floods11"

BUCKET_API_URL = (
    f"https://storage.googleapis.com/storage/v1/b/"
    f"{BUCKET_NAME}/o"
)

params = {
    "maxResults": 1000,
    "fields": "items(name,size,contentType,updated),nextPageToken",
}

print("\n1. BUCKET CONNECTION")
print("-" * 70)

print(f"Bucket       : gs://{BUCKET_NAME}")
print(f"API endpoint : {BUCKET_API_URL}")

# ------------------------------------------------------------
# 2. Request first object page
# ------------------------------------------------------------

try:

    response = requests.get(
        BUCKET_API_URL,
        params=params,
        timeout=60,
    )

    print(f"HTTP status  : {response.status_code}")

    response.raise_for_status()

    bucket_data = response.json()

    print("Connection   : SUCCESS")

except requests.RequestException as exc:

    bucket_data = None

    print("Connection   : FAILED")
    print(f"Error        : {exc}")

# ------------------------------------------------------------
# 3. Parse bucket objects
# ------------------------------------------------------------

if bucket_data is not None:

    objects = bucket_data.get("items", [])

    print("\n2. FIRST PAGE SUMMARY")
    print("-" * 70)

    print(f"Objects returned : {len(objects):,}")
    print(
        f"More pages       : "
        f"{'YES' if bucket_data.get('nextPageToken') else 'NO'}"
    )

    object_records = []

    for obj in objects:

        object_records.append(
            {
                "name": obj.get("name"),
                "size_bytes": int(obj.get("size", 0)),
                "content_type": obj.get("contentType"),
                "updated": obj.get("updated"),
            }
        )

    bucket_df = pd.DataFrame(object_records)

    # --------------------------------------------------------
    # 4. Display first objects
    # --------------------------------------------------------

    print("\n3. FIRST 30 OBJECTS")
    print("-" * 70)

    for name in bucket_df["name"].head(30):
        print(f"  - {name}")

    # --------------------------------------------------------
    # 5. Identify top-level prefixes
    # --------------------------------------------------------

    print("\n4. TOP-LEVEL BUCKET STRUCTURE")
    print("-" * 70)

    top_level_counts = Counter()

    for name in bucket_df["name"]:

        parts = PurePosixPath(name).parts

        if parts:
            top_level_counts[parts[0]] += 1

    for prefix, count in sorted(top_level_counts.items()):
        print(f"{prefix:<40} {count:>8,} objects")

    # --------------------------------------------------------
    # 6. File extension summary
    # --------------------------------------------------------

    print("\n5. FILE EXTENSION SUMMARY")
    print("-" * 70)

    extension_counts = Counter()

    for name in bucket_df["name"]:

        suffix = PurePosixPath(name).suffix.lower()

        if suffix:
            extension_counts[suffix] += 1
        else:
            extension_counts["[NO EXTENSION]"] += 1

    for extension, count in extension_counts.most_common(20):
        print(f"{extension:<20} {count:>8,}")

    # --------------------------------------------------------
    # 7. Save first-page metadata
    # --------------------------------------------------------

    metadata_output = TABLE_DIR / "sen1floods11_bucket_first_page.csv"

    bucket_df.to_csv(
        metadata_output,
        index=False,
    )

    print("\n6. METADATA OUTPUT")
    print("-" * 70)

    print(f"Saved metadata : {metadata_output}")

    print("\n" + "=" * 70)
    print("PUBLIC BUCKET INSPECTION COMPLETE")
    print("=" * 70)

else:

    print("\nBucket inspection could not be completed.")
```

```python
# ============================================================
# CELL 5: BUILD COMPLETE SEN1FLOODS11 BUCKET INVENTORY
# ============================================================

import requests
import pandas as pd
from collections import Counter
from pathlib import PurePosixPath

print("=" * 70)
print("COMPLETE SEN1FLOODS11 BUCKET INVENTORY")
print("=" * 70)

BUCKET_NAME = "sen1floods11"

BUCKET_API_URL = (
    f"https://storage.googleapis.com/storage/v1/b/"
    f"{BUCKET_NAME}/o"
)

# ------------------------------------------------------------
# 1. Retrieve all bucket object metadata using pagination
# ------------------------------------------------------------

print("\n1. RETRIEVING COMPLETE OBJECT INVENTORY")
print("-" * 70)

all_objects = []

page_token = None
page_number = 1

while True:

    params = {
        "maxResults": 1000,
        "fields": (
            "items(name,size,contentType,updated),"
            "nextPageToken"
        ),
    }

    if page_token is not None:
        params["pageToken"] = page_token

    response = requests.get(
        BUCKET_API_URL,
        params=params,
        timeout=120,
    )

    response.raise_for_status()

    page_data = response.json()

    page_objects = page_data.get("items", [])

    all_objects.extend(page_objects)

    print(
        f"Page {page_number:<3} | "
        f"Objects retrieved: {len(page_objects):>5,} | "
        f"Running total: {len(all_objects):>7,}"
    )

    page_token = page_data.get("nextPageToken")

    if page_token is None:
        break

    page_number += 1


# ------------------------------------------------------------
# 2. Convert metadata to DataFrame
# ------------------------------------------------------------

inventory_records = []

for obj in all_objects:

    name = obj.get("name", "")

    path = PurePosixPath(name)

    inventory_records.append(
        {
            "name": name,
            "size_bytes": int(obj.get("size", 0)),
            "size_mb": (
                int(obj.get("size", 0))
                / (1024 ** 2)
            ),
            "content_type": obj.get("contentType"),
            "updated": obj.get("updated"),
            "suffix": path.suffix.lower(),
            "filename": path.name,
            "parent": str(path.parent),
        }
    )

inventory_df = pd.DataFrame(inventory_records)


# ------------------------------------------------------------
# 3. Inventory summary
# ------------------------------------------------------------

print("\n2. COMPLETE INVENTORY SUMMARY")
print("-" * 70)

print(f"Total pages          : {page_number:,}")
print(f"Total bucket objects : {len(inventory_df):,}")

total_size_gb = (
    inventory_df["size_bytes"].sum()
    / (1024 ** 3)
)

print(f"Total object size    : {total_size_gb:,.2f} GB")


# ------------------------------------------------------------
# 4. File extension summary
# ------------------------------------------------------------

print("\n3. FILE TYPE SUMMARY")
print("-" * 70)

extension_summary = (
    inventory_df
    .assign(
        extension=inventory_df["suffix"].replace(
            "",
            "[NO EXTENSION]",
        )
    )
    .groupby("extension")
    .agg(
        object_count=("name", "count"),
        total_size_mb=("size_mb", "sum"),
    )
    .sort_values(
        "object_count",
        ascending=False,
    )
)

print(extension_summary.head(20).to_string())


# ------------------------------------------------------------
# 5. Identify major dataset components
# ------------------------------------------------------------

print("\n4. DATASET COMPONENT DISCOVERY")
print("-" * 70)

component_keywords = {
    "Hand-Labeled Labels": "hand_labeled_label",
    "Hand-Labeled S1": "hand_labeled_s1",
    "Hand-Labeled S2": "hand_labeled_s2",
    "Weak Labels": "weak_label",
    "Weak S1": "weak_s1",
    "Weak S2": "weak_s2",
    "DEM": "dem",
    "Permanent Water": "permanent_water",
    "Metadata": "metadata",
    "Train": "train",
    "Validation": "valid",
    "Test": "test",
}

component_summary = []

for component, keyword in component_keywords.items():

    matches = inventory_df[
        inventory_df["name"]
        .str.contains(
            keyword,
            case=False,
            na=False,
        )
    ]

    component_summary.append(
        {
            "component": component,
            "keyword": keyword,
            "object_count": len(matches),
            "size_mb": matches["size_mb"].sum(),
        }
    )

component_df = pd.DataFrame(component_summary)

print(
    component_df.to_string(
        index=False,
        formatters={
            "size_mb": "{:,.2f}".format,
        },
    )
)


# ------------------------------------------------------------
# 6. GeoTIFF discovery
# ------------------------------------------------------------

print("\n5. GEOTIFF DISCOVERY")
print("-" * 70)

geotiff_df = inventory_df[
    inventory_df["suffix"].isin(
        [".tif", ".tiff"]
    )
].copy()

print(f"GeoTIFF objects : {len(geotiff_df):,}")

if not geotiff_df.empty:

    print("\nFirst 30 GeoTIFF objects:")

    for name in geotiff_df["name"].head(30):
        print(f"  - {name}")


# ------------------------------------------------------------
# 7. CSV and metadata file discovery
# ------------------------------------------------------------

print("\n6. CSV / METADATA FILE DISCOVERY")
print("-" * 70)

metadata_candidates = inventory_df[
    inventory_df["suffix"].isin(
        [
            ".csv",
            ".geojson",
            ".json",
            ".txt",
        ]
    )
].copy()

important_metadata = metadata_candidates[
    metadata_candidates["name"].str.contains(
        (
            "metadata|train|valid|test|split|"
            "catalog"
        ),
        case=False,
        regex=True,
        na=False,
    )
]

print(
    f"Potential metadata files : "
    f"{len(important_metadata):,}"
)

print("\nFirst 50 potential metadata files:")

for name in important_metadata["name"].head(50):
    print(f"  - {name}")


# ------------------------------------------------------------
# 8. Save complete inventory
# ------------------------------------------------------------

inventory_output = (
    TABLE_DIR
    / "sen1floods11_complete_bucket_inventory.csv"
)

inventory_df.to_csv(
    inventory_output,
    index=False,
)

component_output = (
    TABLE_DIR
    / "sen1floods11_component_summary.csv"
)

component_df.to_csv(
    component_output,
    index=False,
)

print("\n7. OUTPUT FILES")
print("-" * 70)

print(f"Complete inventory : {inventory_output}")
print(f"Component summary  : {component_output}")


print("\n" + "=" * 70)
print("COMPLETE BUCKET INVENTORY FINISHED")
print("=" * 70)
```

```python
# ============================================================
# CELL 6: DISCOVER ACTUAL SEN1FLOODS11 DATA STRUCTURE
# ============================================================

from pathlib import PurePosixPath
from collections import Counter
import pandas as pd

print("=" * 70)
print("SEN1FLOODS11 DATA STRUCTURE DISCOVERY")
print("=" * 70)

# ------------------------------------------------------------
# 1. Restrict inventory to actual flood-event data
# ------------------------------------------------------------

DATA_PREFIX = "v1.1/data/flood_events/"

flood_data_df = inventory_df[
    inventory_df["name"].str.startswith(DATA_PREFIX)
].copy()

print("\n1. FLOOD EVENT DATA SUMMARY")
print("-" * 70)

print(f"Flood-event objects : {len(flood_data_df):,}")

flood_data_size_gb = (
    flood_data_df["size_bytes"].sum()
    / (1024 ** 3)
)

print(f"Flood-event size    : {flood_data_size_gb:,.2f} GB")


# ------------------------------------------------------------
# 2. Parse bucket path components
# ------------------------------------------------------------

def get_path_part(path_string, position):
    """
    Safely retrieve a component from a POSIX-style object path.
    """

    parts = PurePosixPath(path_string).parts

    if len(parts) > position:
        return parts[position]

    return None


flood_data_df["level_0"] = flood_data_df["name"].apply(
    lambda x: get_path_part(x, 0)
)

flood_data_df["level_1"] = flood_data_df["name"].apply(
    lambda x: get_path_part(x, 1)
)

flood_data_df["level_2"] = flood_data_df["name"].apply(
    lambda x: get_path_part(x, 2)
)

flood_data_df["level_3"] = flood_data_df["name"].apply(
    lambda x: get_path_part(x, 3)
)

flood_data_df["level_4"] = flood_data_df["name"].apply(
    lambda x: get_path_part(x, 4)
)

flood_data_df["level_5"] = flood_data_df["name"].apply(
    lambda x: get_path_part(x, 5)
)


# ------------------------------------------------------------
# 3. Display unique hierarchy
# ------------------------------------------------------------

print("\n2. FLOOD EVENT DIRECTORY HIERARCHY")
print("-" * 70)

for level in [
    "level_3",
    "level_4",
    "level_5",
]:

    print(f"\n{level.upper()}")

    level_counts = (
        flood_data_df[level]
        .value_counts(dropna=False)
    )

    print(level_counts.head(30).to_string())


# ------------------------------------------------------------
# 4. Summarize immediate directories under HandLabeled
# ------------------------------------------------------------

print("\n3. HANDLABELED COMPONENTS")
print("-" * 70)

hand_labeled_df = flood_data_df[
    flood_data_df["name"].str.contains(
        "/HandLabeled/",
        case=False,
        regex=False,
        na=False,
    )
].copy()

print(f"HandLabeled objects : {len(hand_labeled_df):,}")

hand_components = (
    hand_labeled_df
    .groupby("level_4", dropna=False)
    .agg(
        object_count=("name", "count"),
        total_size_mb=("size_mb", "sum"),
    )
    .sort_values(
        "object_count",
        ascending=False,
    )
)

print(
    hand_components.to_string(
        formatters={
            "total_size_mb": "{:,.2f}".format,
        }
    )
)


# ------------------------------------------------------------
# 5. Print sample files from each HandLabeled component
# ------------------------------------------------------------

print("\n4. SAMPLE FILES BY HANDLABELED COMPONENT")
print("-" * 70)

for component in hand_components.index:

    if pd.isna(component):
        continue

    component_files = hand_labeled_df[
        hand_labeled_df["level_4"] == component
    ]

    print("\n" + "-" * 70)
    print(f"COMPONENT: {component}")
    print("-" * 70)

    for name in component_files["name"].head(5):
        print(f"  - {name}")


# ------------------------------------------------------------
# 6. Discover CSV files
# ------------------------------------------------------------

print("\n5. CSV FILE DISCOVERY")
print("-" * 70)

csv_df = inventory_df[
    inventory_df["suffix"] == ".csv"
].copy()

print(f"CSV objects : {len(csv_df):,}")

for name in csv_df["name"]:
    print(f"  - {name}")


# ------------------------------------------------------------
# 7. Discover hand-labeled GeoTIFF naming patterns
# ------------------------------------------------------------

print("\n6. HANDLABELED GEOTIFF NAMING PATTERNS")
print("-" * 70)

hand_tif_df = hand_labeled_df[
    hand_labeled_df["suffix"].isin(
        [".tif", ".tiff"]
    )
].copy()

print(f"HandLabeled GeoTIFFs : {len(hand_tif_df):,}")

# Extract filename stem
hand_tif_df["stem"] = hand_tif_df["filename"].apply(
    lambda x: PurePosixPath(x).stem
)

# Extract final underscore token
hand_tif_df["final_token"] = (
    hand_tif_df["stem"]
    .str.split("_")
    .str[-1]
)

token_summary = (
    hand_tif_df["final_token"]
    .value_counts()
)

print("\nMost common filename suffix tokens:")

print(token_summary.head(30).to_string())


# ------------------------------------------------------------
# 8. Search for one known sample across all data components
# ------------------------------------------------------------

SAMPLE_ID = "Bolivia_103757"

print(
    f"\n7. CROSS-COMPONENT SEARCH: {SAMPLE_ID}"
)
print("-" * 70)

sample_matches = inventory_df[
    inventory_df["name"].str.contains(
        SAMPLE_ID,
        case=False,
        regex=False,
        na=False,
    )
].copy()

print(f"Matching objects : {len(sample_matches):,}")

for name in sample_matches["name"]:
    print(f"  - {name}")


# ------------------------------------------------------------
# 9. Save structure summaries
# ------------------------------------------------------------

hand_component_output = (
    TABLE_DIR
    / "sen1floods11_handlabeled_components.csv"
)

hand_components.reset_index().to_csv(
    hand_component_output,
    index=False,
)

hand_tif_output = (
    TABLE_DIR
    / "sen1floods11_handlabeled_geotiffs.csv"
)

hand_tif_df.to_csv(
    hand_tif_output,
    index=False,
)

print("\n8. OUTPUT FILES")
print("-" * 70)

print(
    f"Component summary : "
    f"{hand_component_output}"
)

print(
    f"HandLabeled TIFFs : "
    f"{hand_tif_output}"
)

print("\n" + "=" * 70)
print("DATA STRUCTURE DISCOVERY COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# CELL 7: DOWNLOAD SEN1FLOODS11 METADATA AND OFFICIAL SPLITS
# ============================================================

import requests
import pandas as pd
from pathlib import Path

print("=" * 70)
print("SEN1FLOODS11 METADATA AND SPLIT ACQUISITION")
print("=" * 70)

# ------------------------------------------------------------
# 1. Create metadata directory
# ------------------------------------------------------------

METADATA_DIR = SEN1_DIR / "metadata"
SPLITS_DIR = SEN1_DIR / "splits"

METADATA_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

SPLITS_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

print("\n1. LOCAL DIRECTORIES")
print("-" * 70)

print(f"Metadata directory : {METADATA_DIR}")
print(f"Splits directory   : {SPLITS_DIR}")


# ------------------------------------------------------------
# 2. Files to download
# ------------------------------------------------------------

metadata_objects = [
    "v1.1/Sen1Floods11_Metadata.geojson",
]

split_objects = [
    "v1.1/splits/flood_handlabeled/flood_bolivia_data.csv",
    "v1.1/splits/flood_handlabeled/flood_test_data.csv",
    "v1.1/splits/flood_handlabeled/flood_train_data.csv",
    "v1.1/splits/flood_handlabeled/flood_valid_data.csv",
]


# ------------------------------------------------------------
# 3. Public bucket download function
# ------------------------------------------------------------

def download_public_object(
    bucket_name,
    object_name,
    output_path,
):
    """
    Download one public Google Cloud Storage object.
    """

    url = (
        f"https://storage.googleapis.com/"
        f"{bucket_name}/{object_name}"
    )

    response = requests.get(
        url,
        timeout=120,
    )

    response.raise_for_status()

    output_path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    output_path.write_bytes(
        response.content
    )

    return {
        "object_name": object_name,
        "output_path": output_path,
        "size_bytes": len(response.content),
    }


# ------------------------------------------------------------
# 4. Download metadata
# ------------------------------------------------------------

print("\n2. DOWNLOADING METADATA")
print("-" * 70)

download_records = []

for object_name in metadata_objects:

    output_path = (
        METADATA_DIR
        / Path(object_name).name
    )

    record = download_public_object(
        BUCKET_NAME,
        object_name,
        output_path,
    )

    download_records.append(record)

    print(
        f"[DOWNLOADED] "
        f"{record['output_path'].name:<40} "
        f"{record['size_bytes'] / 1024:>10,.2f} KB"
    )


# ------------------------------------------------------------
# 5. Download official flood splits
# ------------------------------------------------------------

print("\n3. DOWNLOADING OFFICIAL FLOOD SPLITS")
print("-" * 70)

for object_name in split_objects:

    output_path = (
        SPLITS_DIR
        / Path(object_name).name
    )

    record = download_public_object(
        BUCKET_NAME,
        object_name,
        output_path,
    )

    download_records.append(record)

    print(
        f"[DOWNLOADED] "
        f"{record['output_path'].name:<40} "
        f"{record['size_bytes'] / 1024:>10,.2f} KB"
    )


# ------------------------------------------------------------
# 6. Validate downloads
# ------------------------------------------------------------

print("\n4. DOWNLOAD VALIDATION")
print("-" * 70)

download_df = pd.DataFrame(
    download_records
)

for _, row in download_df.iterrows():

    exists = Path(
        row["output_path"]
    ).exists()

    print(
        f"{Path(row['output_path']).name:<40} "
        f"[{'OK' if exists else 'MISSING'}]"
    )


# ------------------------------------------------------------
# 7. Save download manifest
# ------------------------------------------------------------

manifest_path = (
    TABLE_DIR
    / "sen1floods11_metadata_download_manifest.csv"
)

download_df.to_csv(
    manifest_path,
    index=False,
)

print("\n5. DOWNLOAD MANIFEST")
print("-" * 70)

print(f"Saved : {manifest_path}")

print("\n" + "=" * 70)
print("METADATA AND SPLIT ACQUISITION COMPLETE")
print("=" * 70)
```
